```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Configuración para que los gráficos
%matplotlib inline
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Ignorar advertencias
warnings.filterwarnings('ignore')
```

    <h2 style="color: #1a5276; margin-top: 0; border-bottom: 1px solid #dee2e6; padding-bottom: 10px;">🛠️ Ingesta, normalización y diagnóstico</h2>
    <p style="color: #7f8c8d; font-style: italic; margin-bottom: 20px;">Preparación técnica del dataset y auditoría de calidad de la información.</p>
    <div style="margin-bottom: 15px;">
        <b style="color: #1a5276; font-size: 1.1em;">Procesos:</b>
        <ul style="line-height: 1.6; margin-top: 5px;">
            <li><b>Ingesta:</b> Carga del archivo <code>Life_Expectancy_Data.csv</code> desde el entorno local.</li>
            <li><b>Normalización:</b> Transformación de nombres de columnas a formato <i>snake_case</i> (minúsculas y guiones bajos) para garantizar una sintaxis de código limpia y sin errores de espacios.</li>
            <li><b>Detalle de estructura:</b> Inspección de tipos de datos y generación de una matriz transpuesta para identificar la distribución de valores nulos y registros por variable.</li>
            <li><b>Análisis de correlación:</b> Generación de un <b>Mapa de calor (Heatmap)</b> utilizando el coeficiente de Pearson para cuantificar la fuerza de asociación entre las variables numéricas y validar preliminarmente las hipótesis planteadas.</li>
        </ul>
    </div>
</div>

```python
# Ruta del archivo o nombre
file_path = 'Life_Expectancy_Data.csv'

try:
    df = pd.read_csv(file_path)
    df.columns = df.columns.str.strip()
    
    print(f"✅ ¡Dataset cargado! Dimensiones: {df.shape[0]} registros y {df.shape[1]} columnas.")

except FileNotFoundError:
    print(f"❌ Error: No se encontró el archivo '{file_path}'.")

# Visualización de los primeros 5 registros
df.head()
```

```python
# Conversión de todos los nombres de columnas a minúsculas y reemplazar espacios por guiones bajos
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('__', '_')
df.columns = [col.replace('-', '_') for col in df.columns]

# Visualizar los nombres normalizados
print("✅ Columnas normalizadas a snake_case:")
print(df.columns.tolist())

# Mostramos las primeras filas
df.head()
```

```python
# Verificamos qué tipo de dato es cada columna y cuántos valores "no nulos" hay
print("📋 Estructura del Dataset")
df.info()
```

```python
# Resumen de la composición del dataset
df_composicion = pd.DataFrame({
    'Tipo de Dato': df.dtypes,
    'Registros No Nulos': df.count(),
    'Valores Nulos': df.isnull().sum(),
    'Porcentaje Nulos (%)': (df.isnull().sum() / len(df) * 100).round(2)
})

# Visualizar la transpuesta para una fácil lectura
print("📋 Diagnóstico de Composición del Dataset:")
display(df_composicion.T)
```

```python
# Calcular nulos y filtrar solo los que tienen más de 0
nulos_por_columna = df.isnull().sum()
nulos_solo_faltantes = nulos_por_columna[nulos_por_columna > 0].sort_values(ascending=False)

if not nulos_solo_faltantes.empty:
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x=nulos_solo_faltantes.index, y=nulos_solo_faltantes.values, palette='Blues_r')
    
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}', 
                    (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha = 'center', va = 'center', 
                    xytext = (0, 9), 
                    textcoords = 'offset points',
                    fontsize=10, fontweight='bold')

    plt.title('❌ Distribución de Valores Faltantes por Variable', fontsize=15, pad=20, color='#1a5276')
    plt.xticks(rotation=45)
    plt.ylabel('Cantidad de registros nulos')
    plt.xlabel('Variables')
    sns.despine()
    plt.show()
else:
    print("✅ No se detectaron valores faltantes en el dataset actual.")
```

```python
# Selección de solo las variables numéricas para la correlación
corr_matrix = df.select_dtypes(include=[np.number]).corr()

# Configuración del estilo visual
plt.figure(figsize=(15, 10))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Creación del Heatmap de correlación
sns.heatmap(corr_matrix, 
            mask=mask, 
            annot=True, 
            fmt=".2f", 
            cmap='RdBu_r', 
            center=0,
            linewidths=.5, 
            cbar_kws={"shrink": .8})

plt.title('Matriz de correlación de variables (Dataset OMS)', fontsize=16, color='#1a5276', pad=20)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.regplot(x='schooling', y='life_expectancy', data=df, 
            scatter_kws={'alpha':0.4, 'color':'#1a5276'}, 
            line_kws={'color':'red', 'lw':2})

plt.title('Relación entre años de escolaridad y esperanza de vida', fontsize=15, pad=20)
plt.xlabel('Años de escolaridad', fontsize=12)
plt.ylabel('Esperanza de vida (Años)', fontsize=12)
plt.show()

# Calculamos el valor exacto para comentar en el notebook
print(f"📊 Coeficiente de correlación: {df['schooling'].corr(df['life_expectancy']):.2f}")
```

```python
plt.figure(figsize=(12, 6))
sns.violinplot(x='status', y='income_composition_of_resources', data=df, palette='Blues', inner="quart")

plt.title('Distribución del índice de composición de recursos por estatus', fontsize=15, pad=20)
plt.xlabel('Estatus del país', fontsize=12)
plt.ylabel('Índice de composición de recursos', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='infant_deaths', y='life_expectancy', hue='status', data=df, palette='viridis', alpha=0.6)

plt.title('Impacto de la mortalidad infantil en la esperanza de vida', fontsize=15, pad=20)
plt.xlabel('Muertes infantiles (por cada 1000)', fontsize=12)
plt.ylabel('Esperanza de vida', fontsize=12)
plt.xlim(0, 200)
plt.show()
```
