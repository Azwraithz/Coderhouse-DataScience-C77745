```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Configuración para que los gráficos
%matplotlib inline
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Ignorar advertencias
warnings.filterwarnings('ignore')
```

    <h2 style="color: #1a5276; margin-top: 0; border-bottom: 1px solid #dee2e6; padding-bottom: 10px;">📥 Ingesta de datos y exploración inicial</h2>
    <p style="color: #7f8c8d; font-style: italic; margin-bottom: 20px;">Fase técnica de carga, normalización de variables y diagnóstico de calidad de la información.</p>
    <div style="margin-bottom: 15px;">
        <b style="color: #1a5276; font-size: 1.1em;">Propósito de esta sección:</b>
        <p style="line-height: 1.6; margin: 5px 0 0 15px;">En esta etapa cargamos el dataset oficial de la OMS, eliminamos inconsistencias en los nombres de las columnas (espacios en blanco) y realizamos un diagnóstico preventivo sobre valores faltantes y tipos de datos para asegurar la integridad de las pruebas de hipótesis posteriores.</p>
    </div>
</div>

```python
# Ruta del archivo o nombre
file_path = 'Life_Expectancy_Data.csv'

try:
    df = pd.read_csv(file_path)
    df.columns = df.columns.str.strip()
    
    print(f"✅ ¡Dataset cargado! Dimensiones: {df.shape[0]} registros y {df.shape[1]} columnas.")

except FileNotFoundError:
    print(f"❌ Error: No se encontró el archivo '{file_path}'.")

# Visualización de los primeros 5 registros
df.head()
```

```python
# Verificamos qué tipo de dato es cada columna y cuántos valores "no nulos" hay
print("📋 Estructura del Dataset")
df.info()
```

```python
# Resumen de la composición del dataset
df_composicion = pd.DataFrame({
    'Tipo de Dato': df.dtypes,
    'Registros No Nulos': df.count(),
    'Valores Nulos': df.isnull().sum(),
    'Porcentaje Nulos (%)': (df.isnull().sum() / len(df) * 100).round(2)
})

# Visualizar la transpuesta para una fácil lectura
print("📋 Diagnóstico de Composición del Dataset:")
display(df_composicion.T)
```

<div style="background-color: #f8f9fa; padding: 25px; border-radius: 15px; border-left: 10px solid #1a5276; font-family: 'Segoe UI', sans-serif; color: #2c3e50; max-width: 100%; box-sizing: border-box;">
    <h2 style="color: #1a5276; margin-top: 0; border-bottom: 1px solid #dee2e6; padding-bottom: 10px;">🔍 Diagnóstico de integridad de datos</h2>
    <p style="color: #7f8c8d; font-style: italic; margin-bottom: 20px;">Visualización técnica de la completitud del dataset para la toma de decisiones sobre imputación.</p>
    <div style="margin-bottom: 15px;">
        <b style="color: #1a5276; font-size: 1.1em;">Análisis de faltantes:</b>
        <p style="line-height: 1.6; margin: 5px 0 0 15px;">El siguiente gráfico identifica las variables críticas que presentan ausencia de registros. Este paso es vital para determinar si la falta de datos es aleatoria o si responde a problemas estructurales en ciertos países o años.</p>
    </div>
</div>

```python
# Calcular nulos y filtrar solo los que tienen más de 0
nulos_por_columna = df.isnull().sum()
nulos_solo_faltantes = nulos_por_columna[nulos_por_columna > 0].sort_values(ascending=False)

if not nulos_solo_faltantes.empty:
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x=nulos_solo_faltantes.index, y=nulos_solo_faltantes.values, palette='Blues_r')
    
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}', 
                    (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha = 'center', va = 'center', 
                    xytext = (0, 9), 
                    textcoords = 'offset points',
                    fontsize=10, fontweight='bold')

    plt.title('❌ Distribución de Valores Faltantes por Variable', fontsize=15, pad=20, color='#1a5276')
    plt.xticks(rotation=45)
    plt.ylabel('Cantidad de registros nulos')
    plt.xlabel('Variables')
    sns.despine()
    plt.show()
else:
    print("✅ No se detectaron valores faltantes en el dataset actual.")
```

```python
# Selección de solo las variables numéricas para la correlación
corr_matrix = df.select_dtypes(include=[np.number]).corr()

# Configuración del estilo visual
plt.figure(figsize=(15, 10))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Creación del Heatmap de correlación
sns.heatmap(corr_matrix, 
            mask=mask, 
            annot=True, 
            fmt=".2f", 
            cmap='RdBu_r', 
            center=0,
            linewidths=.5, 
            cbar_kws={"shrink": .8})

plt.title('Matriz de correlación de variables (Dataset OMS)', fontsize=16, color='#1a5276', pad=20)
plt.show()
```
