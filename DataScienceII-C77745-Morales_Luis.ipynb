```python
# Ruta del archivo o nombre
file_path = 'Life_Expectancy_Data.csv'

try:
    df = pd.read_csv(file_path)
    df.columns = df.columns.str.strip()
    
    print(f"✅ ¡Dataset cargado! Dimensiones: {df.shape[0]} registros y {df.shape[1]} columnas.")

except FileNotFoundError:
    print(f"❌ Error: No se encontró el archivo '{file_path}'.")

# Visualización de los primeros 5 registros
df.head()
```

```python
# Conversión de todos los nombres de columnas a minúsculas y reemplazar espacios por guiones bajos
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('__', '_')
df.columns = [col.replace('-', '_') for col in df.columns]

# Visualizar los nombres normalizados
print("✅ Columnas normalizadas a snake_case:")
print(df.columns.tolist())

# Mostramos las primeras filas
df.head()
```

```python
# Verificación de qué tipo de dato tiene cada columna
print("📋 Estructura del dataset")
df.info()
```

```python
# Resumen de la composición del dataset
df_composicion = pd.DataFrame({
    'Tipo de Dato': df.dtypes,
    'Registros No Nulos': df.count(),
    'Valores Nulos': df.isnull().sum(),
    'Porcentaje Nulos (%)': (df.isnull().sum() / len(df) * 100).round(2)
})

# Visualizar la transpuesta para una fácil lectura
print("📋 Diagnóstico de composición del dataset:")
display(df_composicion.T)
```

```python
# Calcular nulos y filtrar solo los que tienen más de 0
nulos_por_columna = df.isnull().sum()
nulos_solo_faltantes = nulos_por_columna[nulos_por_columna > 0].sort_values(ascending=False)

if not nulos_solo_faltantes.empty:
    plt.figure(figsize=(12, 6))
    ax = sns.barplot(x=nulos_solo_faltantes.index, y=nulos_solo_faltantes.values, palette='Blues_r')
    
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}', 
                    (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha = 'center', va = 'center', 
                    xytext = (0, 9), 
                    textcoords = 'offset points',
                    fontsize=10, fontweight='bold')

    plt.title('❌ Distribución de valores faltantes por variable', fontsize=15, pad=20, color='#1a5276')
    plt.xticks(rotation=45)
    plt.ylabel('Cantidad de registros nulos')
    plt.xlabel('Variables')
    sns.despine()
    plt.show()
else:
    print("✅ No se detectaron valores faltantes en el dataset actual.")
```

```python
# Selección de solo las variables numéricas para la correlación
corr_matrix = df.select_dtypes(include=[np.number]).corr()

# Configuración del estilo visual
plt.figure(figsize=(15, 10))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

# Creación del Heatmap de correlación
sns.heatmap(corr_matrix, 
            mask=mask, 
            annot=True, 
            fmt=".2f", 
            cmap='RdBu_r', 
            center=0,
            linewidths=.5, 
            cbar_kws={"shrink": .8})

plt.title('Matriz de correlación de variables (Dataset OMS)', fontsize=16, color='#1a5276', pad=20)
plt.show()
```

```python
# Análisis descriptivo de las variables numéricas
df.describe()
```

```python
# Selección de las variables
variables_clave = ['life_expectancy', 'schooling', 'gdp', 'hiv/aids', 'infant_deaths']

print("📝 Resumen estadístico de variables seleccionadas:")
display(df[variables_clave].describe().T)

# Análisis univariado
print("\n🔍 Diagnóstico de composición (Outliers):")
for col in variables_clave:
    mean_val = df[col].mean()
    median_val = df[col].median()
    std_val = df[col].std()
    
    print(f"--- Variable: {col.upper()} ---")
    print(f"   • Promedio: {mean_val:.2f}")
    print(f"   • Mediana (50%): {median_val:.2f}")
    print(f"   • Desviación Estándar: {std_val:.2f}")
    
    if std_val > mean_val:
        print(f"   ⚠️ Alerta: Alta dispersión detectada en {col}. Hay mucha desigualdad.")
    print("")
```

```python
# Gráfico de boxplot para detectar outliers en las variables seleccionadas
plt.figure(figsize=(15, 5))
sns.boxplot(data=df[variables_clave], palette="Set2")
plt.title("Detección de Outliers en variables clave", fontsize=14, fontweight='bold')
plt.xticks(rotation=15)
plt.show()
```

```python
# Función para análisis de variables categóricas con gráficos
def cat_summary(dataset, cat_cols=None, plot=False, top=10):
    if cat_cols is None:
        cat_cols = dataset.select_dtypes(include=['object', 'category']).columns.tolist()

    for col_name in cat_cols:
        
        n_unique = dataset[col_name].nunique()
        freq_table = pd.DataFrame({
            "Count": dataset[col_name].value_counts(), 
            "Percentage (%)": (dataset[col_name].value_counts(normalize=True) * 100).round(2)
        })

        print("\n" + "═"*60)
        print(f" 📊 ANÁLISIS CATEGÓRICO: {col_name.upper()} ".center(60, "═"))
        print(f" Valores únicos: {n_unique} ".center(60))
        print("-" * 60)
        
        # Selección de TOP de países
        if n_unique > top:
            print(f" Mostrando Top {top} de {n_unique} categorías: ".center(60))
            print(freq_table.head(top))
        else:
            print(freq_table)
        print("═"*60)

        # Generación de gráfico de barras
        if plot:
            plt.figure(figsize=(12, 6))
            sns.set_style("whitegrid")
            
            plot_data = dataset[col_name].value_counts().head(top) if n_unique > top else dataset[col_name].value_counts()
            
            ax = sns.barplot(x=plot_data.index, y=plot_data.values, palette='viridis', edgecolor='black')
            
            title_suffix = f"(Top {top})" if n_unique > top else ""
            plt.title(f'Distribución de {col_name.title()} {title_suffix}', fontsize=15, fontweight='bold', pad=15)
            
            plt.xticks(rotation=45, ha='right', fontsize=11)
            plt.ylabel('Frecuencia (N)', fontsize=12)
            
            for p in ax.patches:
                ax.annotate(f'{int(p.get_height())}', 
                            (p.get_x() + p.get_width() / 2., p.get_height()), 
                            ha='center', va='bottom', fontsize=10, fontweight='bold', xytext=(0, 5),
                            textcoords='offset points')
            
            sns.despine()
            plt.tight_layout()
            plt.show()

cat_summary(df, plot=True)
```

```python
# Importación de funciones para análisis de variables numéricas
from scipy.stats import skew, kurtosis

# Función para análisis de variables numéricas con gráficos
def num_summary(dataset, num_cols=None, plot=False):
    if num_cols is None:
        num_cols = dataset.select_dtypes(include=['number']).columns.tolist()

    for col_name in num_cols:
        sk = skew(dataset[col_name].dropna())
        kt = kurtosis(dataset[col_name].dropna())
        
        print("\n" + "═"*60)
        print(f" 📈 ANÁLISIS NUMÉRICO: {col_name.upper()} ".center(60, "═"))
        
        # Resumen de percentiles
        stats = dataset[col_name].describe(percentiles=[0.05, 0.25, 0.50, 0.75, 0.95])
        print(pd.DataFrame(stats).T)
        
        print("-" * 60)
        print(f"🔹 Asimetría: {sk:.2f} | Curtosis: {kt:.2f}")
        
        if abs(sk) > 1:
            print("⚠️ Alerta: Distribución altamente sesgada.")
        if dataset[col_name].isnull().any():
            print(f"❗ Datos faltantes: {dataset[col_name].isnull().sum()} registros.")
        print("═"*60)

        # Histograma + Boxplot
        if plot:
            fig, (ax_hist, ax_box) = plt.subplots(1, 2, figsize=(15, 5), 
                                                 gridspec_kw={'width_ratios': (0.7, 0.3)})
            
            sns.histplot(dataset[col_name], kde=True, color='#3498db', ax=ax_hist, alpha=0.6)
            ax_hist.set_title(f'Distribución y densidad: {col_name}', fontsize=13, fontweight='bold')
            ax_hist.axvline(dataset[col_name].mean(), color='red', linestyle='--', label='Media')
            ax_hist.axvline(dataset[col_name].median(), color='green', linestyle='-', label='Mediana')
            ax_hist.legend()

            # Detección de Outliers con Boxplot
            sns.boxplot(y=dataset[col_name], color='#2ecc71', ax=ax_box, width=0.4)
            ax_box.set_title('Detección de Outliers', fontsize=13, fontweight='bold')
            
            sns.despine()
            plt.tight_layout()
            plt.show()

num_summary(df, plot=True)
```

```python
# Grafico de dispersión con curva de tendencia (regresión polinómica) para "hiv/aids" vs "life_expectancy", con colores por "status"
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 8))

# Scatter plot con regresión para ver el impacto del VIH
sns.scatterplot(data=df, x="hiv/aids", y="life_expectancy", 
                hue="status", style="status", s=100, alpha=0.6,
                palette={'Developed': '#27ae60', 'Developing': '#e67e22'})

# Curva de tendencia
sns.regplot(data=df, x="hiv/aids", y="life_expectancy", 
            scatter=False, order=2, color="#2c3e50", line_kws={'lw': 3, 'ls': '--'})

plt.annotate('Acantilado sanitario:\nPequeños aumentos en VIH\ndestruyen décadas de progreso.', 
             xy=(5, 55), xytext=(15, 75),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1),
             fontsize=11, fontweight='bold', bbox=dict(boxstyle="round", fc="white", ec="red", alpha=0.9))

plt.title('H3: Impacto crítico del VIH/SIDA en la longevidad\n¿Existe un punto de no retorno?', 
          fontsize=16, fontweight='bold', color='#1a5276')
plt.xlabel('Prevalencia de VIH/SIDA (Muertes por cada 1000 nacidos vivos 0-4 años)', fontsize=12)
plt.ylabel('Esperanza de vida (Años)', fontsize=12)

plt.tight_layout()
plt.show()
```
